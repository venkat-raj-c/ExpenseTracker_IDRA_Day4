```python
# Csv writer function

def writeExpenses(csvWriter):
    while True:
        expenseDate = input("Enter date: ").strip()
        expenseCategory = input("Enter category: ").strip()
        expenseAmount = input("Enter amount: ").strip()
        expenseNote = input("Enter note (optional): ").strip()

        if not expenseDate or not expenseCategory or not expenseAmount:
            print("Date, category, and amount cannot be empty. Please enter them again.")
            continue

        expenseRecord = [
            expenseDate,
            expenseCategory,
            expenseAmount,
            expenseNote
        ]

        csvWriter.writerow(expenseRecord)
        print("Expense saved successfully.")
        return
```

```python
def readExpenses():
    with open("expenses.csv", mode='r', newline='', encoding='utf-8') as csvFile:
        csvReader = csv.reader(csvFile)

        grandTotal = 0
        categoryTotals = {}

        print("Date\tCategory\tAmount\tNote")

        for expenseRow in csvReader:
            grandTotal += int(expenseRow[2])

            categoryTotals[expenseRow[1]] = (
                categoryTotals.get(expenseRow[1], 0)
                + int(expenseRow[2])
            )

            print(
                f"{expenseRow[0]}\t"
                f"{expenseRow[1]}\t"
                f"{expenseRow[2]}\t"
                f"{expenseRow[3]}"
            )

        print("Total Amount: ", grandTotal)

        for categoryName, categoryTotal in categoryTotals.items():
            print(categoryName, " : ", categoryTotal)
```

```python
import csv

try:
    with open("expenses.csv", mode='a', newline='', encoding='utf-8') as csvFile:
        csvWriter = csv.writer(csvFile)

        while True:
            print("\nMenu")
            print("1. Enter expense")
            print("2. View expenses")
            print("q. Quit")

            userChoice = input("Enter your choice: ").strip()

            if userChoice == "1":
                writeExpenses(csvWriter)
                csvFile.flush()

            elif userChoice == "2":
                csvFile.seek(0)
                readExpenses()
                csvFile.seek(0, 2)

            elif userChoice.lower() == 'q':
                break

            else:
                print("Invalid choice. Please enter 1 or 2.")

except Exception as error:
    print("File has not been opened properly. Try again", error)
```
